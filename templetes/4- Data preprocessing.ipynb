### 1. Importar librerias requeridas.

```python
# Importamos las librerias utilizadas
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

### 2. Lectura de datos CSV.

```python
# Leemos los dos archivos CSV
dataset = pd.read_csv("datasets/Data.csv") # DS DataFrame
```

### 3. Expororacion de dataset.

```python
# Exploramos filas del conjunto de noticias verdaderas
dataset 
```

```python
X = dataset.iloc[:, :-1].values # iloc Recorta las posiciones indicadas [filas | columnas] y los convierte en un array con .values
y = dataset.iloc[:, 3].values
print("Data X: ", X)
print("Data y: ",y)
```

### 4. Tratamientos de los NAs.

```python
from sklearn.impute import SimpleImputer
```

```python
imputer = SimpleImputer(missing_values=np.nan, strategy="mean")
imputer = imputer.fit(X[:, 1:3]) # Aprende y guarda los parámetros necesarios para trabajar después.
X[:, 1:3] = imputer.transform(X[:, 1:3]) # Usa lo aprendido para reemplazar los NaN.
print(X)
```

### 5. Datos categoricos.

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, LabelEncoder
```

```python
# Indicamos que aplique OneHotEncoder a la columna 0 
# 'remainder="passthrough"' asegura que el resto de las columnas no se modifiquen
ct = ColumnTransformer(
    transformers=[('encoder', OneHotEncoder(), [0])], 
    remainder='passthrough'
)

# Ajustamos y transformamos la matriz X, convirtiéndola directamente a un array de NumPy
X = ct.fit_transform(X)

# Creamos el codificador para la variable dependiente
labelencoder_y = LabelEncoder()

# Ajustamos y transformamos la variable y
y = labelencoder_y.fit_transform(y)

print("Data X:", X)
print("Data y:", y)
```

### 6. Dividir dataset en dataset para entrenamiento y dataset para test.

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, 
    y,
    test_size=0.20,
    random_state=42,
    shuffle=True
)
```
